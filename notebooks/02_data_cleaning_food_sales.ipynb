```python
import pandas as pd

df_sales = pd.read_csv("food_sales_2016_2023.csv")  # Load Source 2 into a DataFrame
```

```python
df_sales.shape  # Check the number of rows and columns
```

```python
df_sales.columns.tolist()  # Display all Source 2 column names
```

```python
df_sales["product group"].tolist()  # Display all 39 food categories
```

```python
df_sales.isna().sum()  # Count missing values in each column
```

```python
df_sales.dtypes  # Check whether sales-value columns are numeric
```

```python
sales_cols = ["product group"] + [
    col for col in df_sales.columns
    if col.endswith(tuple(str(year) for year in range(2018, 2024)))
]  # Keep the category column and sales data from 2018–2023
```

```python
sales_analysis = df_sales[sales_cols].copy()  # Create the 2018–2023 analysis dataset
```

```python
sales_analysis.shape  # Check the resulting structure
```

```python
sales_long = sales_analysis.melt(
    id_vars="product group",
    var_name="Measure_Year",
    value_name="Sales"
)  # Convert the yearly columns into rows
```

```python
sales_long.head()  # Inspect the reshaped sales data
```

```python
sales_long[["Measure", "Year"]] = sales_long["Measure_Year"].str.rsplit(" ", n=1, expand=True)
# Split "current prices 2018" into Measure = "current prices" and Year = "2018"
```

```python
sales_long.head()  # Check that Measure and Year were separated correctly
```

```python
sales_long["Year"] = sales_long["Year"].astype(int)  # Convert Year from text to integer
```

```python
sales_long.dtypes  # Check the data types
```

```python
sales_long = sales_long.drop(columns=["Measure_Year"])  # Remove the temporary combined column
```

```python
sales_long.head()  # Check the final Source 2 structure
```

```python
sales_long.duplicated(
    subset=["product group", "Year", "Measure"]
).sum()  # Check for duplicate category-year-measure records
```

```python
sales_long["Measure"].value_counts()  # Count observations for each sales measure
```

```python
sales_long.to_csv(
    "food_sales_clean.csv",
    index=False
)  # Save the cleaned Source 2 dataset
```
