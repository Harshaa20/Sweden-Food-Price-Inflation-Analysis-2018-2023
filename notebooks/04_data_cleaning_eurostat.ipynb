```python
import pandas as pd

df_eu = pd.read_csv(
    "eurostat_hicp_food.csv"
)  # Load the Eurostat HICP food data
```

```python
df_eu["TIME_PERIOD"].min(), df_eu["TIME_PERIOD"].max()
# Check the earliest and latest available periods
```

```python
df_eu["Unit of measure"].unique()  # Check what the HICP values represent
```

```python
eu_clean = df_eu[
    ["geo", "Geopolitical entity (reporting)", "TIME_PERIOD", "OBS_VALUE"]
].copy()  # Keep only the fields needed for Sweden vs EU analysis
```

```python
eu_clean.head()  # Check the simplified structure
```

```python
eu_clean = eu_clean.rename(columns={
    "Geopolitical entity (reporting)": "Country",
    "TIME_PERIOD": "Month",
    "OBS_VALUE": "HICP"
})  # Rename columns to simple analytical names
```

```python
eu_clean.head()  # Verify the renamed columns
```

```python
eu_clean["Month"] = pd.to_datetime(
    eu_clean["Month"] + "-01"
)  # Convert 2018-01 → 2018-01-01
```

```python
eu_clean.dtypes  # Check the data types
```

```python
eu_clean = eu_clean[
    (eu_clean["Month"].dt.year >= 2018) &
    (eu_clean["Month"].dt.year <= 2023)
].copy()  # Keep only the 2018–2023 project period
```

```python
eu_clean.shape  # Check the remaining observations
```

```python
eu_clean = eu_clean[
    eu_clean["Country"].isin([
        "Sweden",
        "European Union - 27 countries (from 2020)"
    ])
].copy()  # Keep only Sweden and the EU-27 benchmark
```

```python
eu_clean.shape  # Check the remaining observations
```

```python
eu_clean.isna().sum()  # Check for missing values in the cleaned Eurostat data
```

```python
eu_clean.duplicated(
    subset=["geo", "Month"]
).sum()  # Check for duplicate geography-month observations
```

```python
eu_clean.to_csv(
    "eurostat_hicp_food_clean.csv",
    index=False
)  # Save the cleaned Eurostat dataset
```
