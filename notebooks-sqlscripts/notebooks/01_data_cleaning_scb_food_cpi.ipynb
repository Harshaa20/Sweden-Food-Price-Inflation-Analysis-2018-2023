```python
df.iloc[:, 1:] = df.iloc[:, 1:].astype(float)  # Convert all monthly CPI columns to float
```

```python
df.dtypes.head()  # Verify the data types after conversion
```

```python
df = pd.concat([
    df[["Product group"]],              # Keep category names as text
    df.iloc[:, 1:].apply(pd.to_numeric) # Convert all monthly CPI values to numbers
], axis=1)
```

```python
df.dtypes.head()  # Verify the data types
```

```python
df["Product group"].duplicated().sum()  # Count duplicate food-category rows
```

```python
df.columns[1:].tolist()[:3], df.columns[1:].tolist()[-3:]  # Check first and last month
```

```python
analysis_cols = ["Product group"] + [
    col for col in df.columns if col[:4].isdigit() and int(col[:4]) <= 2023
]  # Keep Product group plus monthly data from 2018 through 2023
```

```python
df_analysis = df[analysis_cols].copy()  # Create a separate analysis dataset"
```

```python
df_analysis.shape  # Check rows and columns after filtering
```

```python
detailed_analysis = df_analysis[
    df_analysis["Product group"].str.split(" ").str[0].str.count(r"\.") == 3
].copy()  # Keep only the 60 most detailed food categories
```

```python
detailed_analysis.shape  # Confirm the number of detailed categories and months
```

```python
cpi_long = detailed_analysis.melt(
    id_vars="Product group",
    var_name="Month",
    value_name="CPI"
)  # Convert monthly columns into rows: one category-month observation per row
```

```python
cpi_long.shape  # Check the new number of rows and columns
```

```python
cpi_long.head()  # Preview the first few category-month observations
```

```python
cpi_long["Product group"] = (
    cpi_long["Product group"]
    .str.replace(r"^\S+\s+", "", regex=True)  # Remove the classification code
    .str.replace(" (ND)", "", regex=False)    # Remove the ND label
    .str.strip()                              # Remove extra spaces
)
```

```python
cpi_long.head()  # Check the cleaned category names
```

```python
cpi_long.isna().sum()  # Count missing values in each column
```

```python
cpi_long[cpi_long["CPI"].isna()][["Product group", "Month"]]  # Show every category-month with missing CPI
```

```python
cpi_long.groupby("Product group")["CPI"].apply(lambda x: x.isna().sum()).sort_values(ascending=False).head()
# Count missing CPI observations for each food category
```

```python
cpi_long = cpi_long.dropna(subset=["CPI"]).copy()  # Remove rows where CPI is missing
```

```python
cpi_long.shape  # Check the remaining observations
```

```python
cpi_long.duplicated(subset=["Product group", "Month"]).sum()  # Check for duplicate category-month observations
```

```python
cpi_long["Month"] = pd.to_datetime(
    cpi_long["Month"].str.replace("M", "-") + "-01"
)  # Convert 2018M01 → 2018-01-01
```

```python
cpi_long.dtypes  # Check the data types after conversion
```

```python
cpi_long["CPI"].describe()  # Check basic statistics and the overall CPI range
```

```python
cpi_long.to_csv("scb_food_cpi_clean.csv", index=False)  # Save the cleaned CPI dataset
```
