```python
df_import = pd.read_csv(
    "scb_import_price_index_updated.csv",
    skiprows=1,
    na_values=".."
)  # Skip the title row and treat ".." as missing
```

```python
import_cols = ["products by SPIN 2015"] + [
    col for col in df_import.columns
    if col[:4].isdigit() and int(col[:4]) <= 2023
]  # Keep the product column and monthly data from 2018–2023
```

```python
import_analysis = df_import[import_cols].copy()  # Create the 2018–2023 analysis dataset
```

```python
import_analysis.shape  # Check the resulting structure
```

```python
import_long = import_analysis.melt(
    id_vars="products by SPIN 2015",
    var_name="Month",
    value_name="Import_Price_Index"
)  # Convert monthly columns into rows
```

```python
import_long.shape  # Check the new structure
```

```python
import_long["Month"] = pd.to_datetime(
    import_long["Month"].str.replace("M", "-") + "-01"
)  # Convert 2018M01 → 2018-01-01
```

```python
import_long.dtypes  # Verify the data types
```

```python
import_long["Month"].duplicated().sum()  # Check for duplicate months
```

```python
import_long["Import_Price_Index"].describe()  # Check basic statistics of the import-price index
```

```python
import_long = import_long.rename(
    columns={"products by SPIN 2015": "Product_Group"}
)  # Rename the SCB product column
```

```python
import_long.head()  # Verify the renamed column
```

```python
import_long.to_csv(
    "scb_import_price_clean.csv",
    index=False
)  # Save the cleaned Source 3 dataset
```
